```python
#NumPy

import numpy as np

#Simple array

marks = np.array([95,87,72,68,91])
print(marks)
print(type(marks))

#Basic operations -- yeh sab ek line me aa jayenge

print(np.mean(marks)) #Average
print(np.max(marks)) #Maximum
print(np.min(marks)) #Minimum
print(np.sum(marks)) #Sum
```

```python
#Pandas

import pandas as pd

#Make a dataframe manually

student_data = {
    "Name" : ["Rahul", "Priya", "Arjun", "Sneha"],
    "Marks" : [87,92,78,95],
    "Attendance" : [85,90,70,88],
    "Passed" : [True,True,True,True]
}

df = pd.DataFrame(student_data)

print(df)

print(df.shape) #Rows and columns
print(df.head(2)) #First 2 rows
print(df.info()) #Data types and missing values
print(df.describe()) #Statistical summary

#Specific column access

print(df["Marks"])

#Filter -- Only those students whos marks are more than 90

print(df[df["Marks"]>90])
```

```python
#Matplotlib

import matplotlib.pyplot as plt

#Simple line plot

marks = [87,92,78,95,88]
students = ["Rahul", "Priya", "Arjun", "Sneha", "Rohan"]

plt.plot(students, marks, marker = 'o', color='blue')
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#Bar chart

plt.bar(students, marks, color='red')
plt.title("Student Marks Comparison")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#Scatter plot -- relationship between 2 features

attendance = [85,90,70,88,92]
marks = [87,92,78,95,88]

plt.scatter(attendance, marks, color='green', s = 100)
plt.title("Attendance vs Marks")
plt.xlabel("Attendance %")
plt.ylabel("Marks")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#Step 1 -- data making (In real, there will be a CSV file)

data = {
    "Attendance" : [85,90,70,88,92,65,78,95],
    "Study_hours" : [5,7,3,6,8,2,4,9],
    "Marks" : [82,92,78,95,88,60,75,98]

}

df = pd.DataFrame(data)


#Step 2 -- Explore the data

print(df.describe())
print(df.isnull().sum())


#Step 3 -- Calculation with NumPy

print("Average Marks :", np.mean(df["Marks"]))
print("Maximum Marks :", np.max(df["Marks"]))


#Step 4 -- Data Visualization

plt.scatter(df["Attendance"], df["Marks"], color = 'blue', s = 100)
plt.title("Attendance vs Marks")
plt.xlabel("Attendance %")
plt.ylabel("Marks")
plt.show()
```
